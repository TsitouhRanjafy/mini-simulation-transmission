```python
nombre = 255

# Integer vers bits
bits = format(nombre, '08b')

# Bits vers integer
resultat = int(bits, 2)

print(bits)      # 11111111
print(resultat)  # 255
```

## Text to code ASCII

```python
texte = "Tsitohaina"
ascii = ' '.join(str(ord(caractere)) for caractere in texte)
print(ascii)
```

## Text to Binary


```python
Text = "Tsi"
```

```python
b = ' '.join(format(ord(char), '08b') for char in Text)
print(b)
```

## Text to UTF-8 Binary

```python
b = ' '.join(format(b, '08b') for b in bytearray(Text, 'utf-8'))
print(b)
```

```python
# Text string
text = "Hello World"
 
# Encode text to binary data
data = text.encode("utf-8") 
 
# Decode binary data back to text string
text = data.decode("utf-8")  

# data is not a str, it's a bytes
print(data)
print(type(data))
print(list(data)) 

# afficher le bits réels
print([format(byte, '08b') for byte in data])

print(text)
```

## Binary to Text

```python
b = '01010100011100110110100101110100011011110110100001100001011010010110111001100001'
s = ''.join(chr(int(b[i:i+8],2)) for i in range(0, len(b), 8))
print(s)
```

```python
import codecs
b = '01010100011100110110100101110100011011110110100001100001011010010110111001100001'
hex_string = hex(int(b, 2))[2:]

if len(hex_string) % 2 != 0:
    hex_string = '0' + hex_string
s = codecs.decode(hex_string, 'hex').decode()
print(s)
```
